```python
import numpy as np
import time

# Python list approach
lst = list(range(1_000_000))
t0 = time.time()
result = [x * 2 for x in lst]
print(f"List:  {(time.time()-t0)*1000:.1f} ms")

# NumPy approach
arr = np.arange(1_000_000)
t0 = time.time()
result = arr * 2
print(f"NumPy: {(time.time()-t0)*1000:.1f} ms")
```
