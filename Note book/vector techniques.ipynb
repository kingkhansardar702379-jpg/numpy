```python
# =====================================================
# AI NLP PROJECT
# ONE HOT ENCODING
# BAG OF WORDS
# TF-IDF
# LARGE PARAGRAPH EXAMPLE
# =====================================================

import pandas as pd

from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# =====================================================
```

```python
# STEP 1 : LARGE PARAGRAPH
# =====================================================

paragraph = """
Artificial Intelligence is transforming the modern world.
AI is used in healthcare, education, banking, cybersecurity,
business, transportation, and many other industries.

Machine Learning is a branch of Artificial Intelligence.
It allows computers to learn from data and improve their
performance without being explicitly programmed.

Deep Learning is a subset of Machine Learning that uses
neural networks to solve complex problems. Natural Language
Processing helps computers understand human language.

AI technologies are used in chatbots, recommendation systems,
virtual assistants, image recognition, speech recognition,
and self-driving cars.

Many organizations use Artificial Intelligence to automate
tasks, analyze large amounts of data, and make better decisions.
AI improves productivity, efficiency, and innovation.

In the future, Artificial Intelligence is expected to play
an even bigger role in solving real-world problems and
improving human life.
"""

print("\n")
print("=" * 60)
print("ORIGINAL PARAGRAPH")
print("=" * 60)

print(paragraph)

# =====================================================
```

```python
# STEP 2 : ONE HOT ENCODING
# =====================================================

print("\n")
print("=" * 60)
print("STEP 2 : ONE HOT ENCODING")
print("=" * 60)

# Convert paragraph into words

words = paragraph.lower().split()

# Remove duplicates

unique_words = list(set(words))

# Create dataframe

df_words = pd.DataFrame(unique_words, columns=["Words"])

# One Hot Encoder

encoder = OneHotEncoder(sparse_output=False)

one_hot_matrix = encoder.fit_transform(df_words)

print("\nShape of Matrix:")
print(one_hot_matrix.shape)

print("\nFirst 10 Rows:")
print(one_hot_matrix[:10])

# Save CSV

one_hot_df = pd.DataFrame(one_hot_matrix)

one_hot_df.to_csv(
    "OneHotEncoding_Output.csv",
    index=False
)

print("\nOne Hot Encoding CSV Saved")
```

```python
# STEP 3 : BAG OF WORDS
# =====================================================

print("\n")
print("=" * 60)
print("STEP 3 : BAG OF WORDS")
print("=" * 60)

documents = [paragraph]

bow = CountVectorizer()

bow_matrix = bow.fit_transform(documents)

print("\nVocabulary:\n")

print(bow.get_feature_names_out())

print("\nBoW Matrix:\n")

print(bow_matrix.toarray())

# Save CSV

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=bow.get_feature_names_out()
)

bow_df.to_csv(
    "BagOfWords_Output.csv",
    index=False
)

print("\nBag Of Words CSV Saved")
```

```python
# STEP 4 : TF-IDF
# =====================================================

print("\n")
print("=" * 60)
print("STEP 4 : TF-IDF")
print("=" * 60)

tfidf = TfidfVectorizer()

tfidf_matrix = tfidf.fit_transform(documents)

print("\nVocabulary:\n")

print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:\n")

print(tfidf_matrix.toarray())

# Save CSV

tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

tfidf_df.to_csv(
    "TFIDF_Output.csv",
    index=False
)

print("\nTF-IDF CSV Saved")
```

```python
# STEP 5 : TOP IMPORTANT WORDS
# =====================================================

print("\n")
print("=" * 60)
print("TOP IMPORTANT WORDS")
print("=" * 60)

scores = zip(
    tfidf.get_feature_names_out(),
    tfidf_matrix.toarray()[0]
)

sorted_scores = sorted(
    scores,
    key=lambda x: x[1],
    reverse=True
)

for word, score in sorted_scores[:20]:
    print(word, "=", round(score, 4))

# =====================================================
```

```python
# STEP 6 : PROJECT COMPLETE
# =====================================================

print("\n")
print("=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nGenerated Files:")

print("1. OneHotEncoding_Output.csv")
print("2. BagOfWords_Output.csv")
print("3. TFIDF_Output.csv")
```
